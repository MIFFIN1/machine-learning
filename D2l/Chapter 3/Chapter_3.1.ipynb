```python
%matplotlib inline
import math
import time
import numpy as np
import torch
from d2l import torch as d2l

# Vectorization for Speed

n = 10000
a = torch.ones(n)
b = torch.ones(n)

# For loops implementation to add vectors

c = torch.zeros(n)
t = time.time()
for i in range(n):
  c[i] = a[i] + b[i]
time1 = f'{time.time() -t:.5f} sec'
print(time1)

# Mathematical method to add vectors

t = time.time()
d = a + b
time2 = f'{time.time()- t:.5f} sec'
print(time2)
```

```python
# Normal Distrubtion

def normal (x, mu, sigma):
  p = 1 / math.sqrt(2 * math.pi * sigma**2)
  return p * np.exp(-0.5 * (x - mu)**2 / sigma**2)

  # NumPy for visualization
  x = np.arange(-7,7,0.01)

  # Mean and STD
  params = [(0, 1), (0, 2), (3, 1)]
  d2l.plot(x, [normal(x, mu, sigma) for mu, sigma in params], xlabel='x',
         ylabel='p(x)', figsize=(4.5, 2.5),
         legend=[f'mean {mu}, std {sigma}' for mu, sigma in params])
```
